```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import PolynomialFeatures

# Dataset
X = np.array([1, 2, 3, 4, 5]).reshape(-1, 1)
y = np.array([2, 5, 10, 17, 26])

# 1. Plot the original data
plt.scatter(X, y)
plt.xlabel("Years of Experience")
plt.ylabel("Salary (Lakhs)")
plt.title("Experience vs Salary")
plt.show()

# 2. Create polynomial features
poly = PolynomialFeatures(degree=2)
X_poly = poly.fit_transform(X)

# 3. Train the model
model = LinearRegression()
model.fit(X_poly, y)

# 4. Predict salary
y_pred = model.predict(X_poly)

# 5. Print coefficients and predictions
print("Coefficients:", model.coef_)
print("Intercept:", model.intercept_)
print("Predicted Salary:", model.predict(poly.transform([[6]])))

# 6. Plot the polynomial curve
plt.scatter(X, y, label="Actual Data")
plt.plot(X, y_pred, color="red", label="Polynomial Curve")
plt.xlabel("Years of Experience")
plt.ylabel("Salary (Lakhs)")
plt.legend()
plt.show()
```
