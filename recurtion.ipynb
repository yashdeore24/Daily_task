```python
def factorial(n):
    if n == 0 or n == 1:
        return 1
    return n * factorial(n - 1)

num = int(input())
print(factorial(num))
```

```python
# Write your code here

class Employee:
    def __init__(self, name):
        self.name = name

    def show_role(self):
        print(self.name, " - Employee")


class Manager(Employee):
    def show_role(self):
        print(self.name, " - Manager")


name = input()
obj = Manager(name)
obj.show_role()
```
